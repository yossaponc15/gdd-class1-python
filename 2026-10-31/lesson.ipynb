## 1️⃣ แบบเดิม: dict + function

ข้อมูลอยู่ใน dict · การกระทำอยู่ในฟังก์ชัน · ต้องส่ง `robot` เข้าไปทุกครั้ง

```python
def create_robot(name, battery=100):
    return {"name": name, "battery": battery}

def move(robot, distance):
    robot["battery"] -= distance // 10
    print(f"{robot['name']} เดิน {distance} m เหลือ {robot['battery']}%")

bolt = create_robot("Bolt")
move(bolt, 50)
move(bolt, 120)
```

> 🤔 ปัญหา: ข้อมูลกับฟังก์ชันแยกกันอยู่ · ใครจะส่ง dict อะไรเข้า `move()` ก็ได้ · โปรแกรมใหญ่ขึ้นจะหาไม่เจอว่าฟังก์ชันไหนใช้กับอะไร

## 2️⃣ แบบใหม่: class = แม่พิมพ์ที่รวม **ข้อมูล + การกระทำ** ไว้ด้วยกัน

| dict + function | class |
|---|---|
| `create_robot(name)` | `__init__(self, name)` |
| `robot["battery"]` | `self.battery` |
| `move(robot, 50)` | `bolt.move(50)` |

```python
class Robot:
    def __init__(self, name, battery=100):    # ทำงานอัตโนมัติตอนสร้าง object
        self.name = name                       # self = object ตัวที่กำลังถูกสร้าง/ใช้งาน
        self.battery = battery

    def move(self, distance):                  # method = function ที่อยู่ใน class
        self.battery -= distance // 10
        print(f"{self.name} เดิน {distance} m เหลือ {self.battery}%")

bolt = Robot("Bolt")
atom = Robot("Atom", 60)
bolt.move(50)
atom.move(120)
print(bolt.battery, atom.battery)              # แต่ละตัวมีข้อมูลของตัวเอง
```

### 🔍 `self` คืออะไร?
`bolt.move(50)` จริง ๆ แล้ว Python แปลงเป็น `Robot.move(bolt, 50)` → `self` ก็คือ `bolt` นั่นเอง

```python
Robot.move(bolt, 30)      # ผลเหมือน bolt.move(30) ทุกประการ
```

## 3️⃣ `__str__`: กำหนดว่า `print(object)` จะแสดงอะไร

```python
print(bolt)               # ยังไม่มี __str__ → ได้ข้อความแปลก ๆ <__main__.Robot object at 0x...>

class Robot:
    def __init__(self, name, battery=100):
        self.name = name
        self.battery = battery

    def move(self, distance):
        self.battery -= distance // 10

    def __str__(self):
        return f"🤖 {self.name} [{self.battery}%]"     # ต้อง return string

bolt = Robot("Bolt")
bolt.move(50)
print(bolt)
```

## 4️⃣ list ของ object

ใช้ท่าเดิมของ list + loop + ตัวสะสม เปลี่ยนแค่ `r["battery"]` ➜ `r.battery`

```python
team = [Robot("Bolt", 80), Robot("Atom", 35), Robot("Cog", 90)]

total = 0
for r in team:
    print(r)
    total += r.battery
print("รวม", total)

low = [r.name for r in team if r.battery < 50]
print("แบตต่ำ:", low)
```

## 5️⃣ Inheritance: สร้าง class ลูกจาก class แม่

- `class CleanerBot(Robot):` → ได้ทุกอย่างของ `Robot` มาฟรี
- `super().__init__(...)` → ให้แม่ตั้งค่าส่วนของแม่ก่อน แล้วลูกค่อยเพิ่มของตัวเอง
- **override** → เขียน method ชื่อเดิมใหม่ในลูก

```python
class Robot:
    def __init__(self, name, battery=100):
        self.name = name
        self.battery = battery

    def work(self):
        return f"{self.name} ทำงานทั่วไป"

    def __str__(self):
        return f"🤖 {self.name} [{self.battery}%]"


class CleanerBot(Robot):
    def __init__(self, name, room, battery=100):
        super().__init__(name, battery)       # ให้ Robot ตั้ง name, battery
        self.room = room                       # ของเพิ่มเฉพาะลูก

    def work(self):                            # override
        return f"{self.name} ทำความสะอาดห้อง {self.room}"


mop = CleanerBot("Mop", "Lab-2")
print(mop)              # ใช้ __str__ ของแม่
print(mop.work())       # ใช้ work ของลูก
```

### 🎭 Polymorphism: เรียก method ชื่อเดียวกัน แต่แต่ละ class ทำต่างกัน

```python
team = [Robot("Bolt"), CleanerBot("Mop", "Lab-2"), CleanerBot("Dusty", "Hall")]
for r in team:
    print(r.work())     # ไม่ต้องเช็กว่าเป็น class ไหน — Python เลือก method ให้เอง
```
