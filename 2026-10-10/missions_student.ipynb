## Level 7: โมดูลคำนวณ 🧮

**วัด:** functions, return หลายค่า, default, function เรียก function

### ภารกิจ
สร้างโมดูลคำนวณเส้นทาง

### ข้อกำหนด
- `distance(p1, p2)` → return ระยะ Manhattan ระหว่างพิกัด 2 มิติ (tuple)
- `route_length(points)` → return ระยะรวมเมื่อเดินตามลำดับจุด (เรียก `distance`)
- `plan(points, speed=2)` → return `(ระยะรวม, เวลา)` โดย เวลา = ระยะ / speed ปัด 1 ตำแหน่ง

### Expected Output
```text
7
18
(18, 9.0)
ระยะ 18 เวลา 6.0
```

### 💡 คำใบ้ (เปิดทีละขั้น ลองคิดเองก่อนนะ!)

<details><summary>ใบ้ 1: เริ่มยังไง</summary>

Manhattan distance = `abs(x1 - x2) + abs(y1 - y2)` · เข้าถึงพิกัดด้วย `p1[0]`, `p1[1]`

</details>

<details><summary>ใบ้ 2: ใช้เครื่องมืออะไร</summary>

`route_length` วนตั้งแต่ index 1: ระยะช่วงที่ i คือ `distance(points[i - 1], points[i])` — ต้อง **เรียกใช้** function เดิม ห้ามเขียนสูตรซ้ำ

</details>

<details><summary>ใบ้ 3: เกือบเฉลย</summary>

return หลายค่า: `return total, round(total / speed, 1)` · `speed=2` คือ default parameter

</details>
