## รอบ 2: Data Structures 📦

```python
# ทายก่อน:
a = [5, 2, 8, 2]
a.append(1)
a.remove(2)
print(a, sorted(set(a)))
print([x * 10 for x in a if x % 2 == 0])
```

```python
# ทายก่อน:
parts = {"gear": 3, "bolt": 10}
parts["gear"] += 2
parts["chip"] = parts.get("chip", 0) + 1
print(parts)
print(sorted(parts.items(), key=lambda p: -p[1]))
```

## รอบ 3: Functions + Errors 🧩

```python
# ทายก่อน:
def power(base, exp=2):
    return base ** exp

print(power(3), power(2, 10), power(exp=3, base=2))
```

```python
# ทายก่อน:
def safe_div(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "หาร 0 ไม่ได้"
    finally:
        print("เช็ก", end=" ")

print(safe_div(9, 3))
print(safe_div(1, 0))
```
